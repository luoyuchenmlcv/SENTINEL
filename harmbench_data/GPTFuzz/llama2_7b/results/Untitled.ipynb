```python
import json
import numpy as np
pth = "vicuna_7b_v1_5-Copy.json"
data = json.load(open(pth, "r"))
```

```python
new_data = dict()

for d in data:
    local_data = data[d]

    if type(local_data[0]['test_case'])==list:
        local_data[0]['test_case'] = local_data[0]['test_case'][0]
    # print(type(data[d]), type(data[d][0]['test_case']))
    new_data[d] = local_data
        # break

with open("vicuna_7b_v1_5_new.json", "w") as f:
    json.dump(new_data, f, indent=4)
```
